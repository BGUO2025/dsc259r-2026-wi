```python
from numpy import genfromtxt
my_data = genfromtxt('/content/gdrive/My Drive/Mall_Customers.csv', delimiter=',', dtype=str)
```

```python
my_data
```

```python
# extract just the income which is the 4th column (index 3) of the 2d array above.
# Because of how it's organized, we'll use iteration to extract this column
incomes = [row[3] for row in my_data]
```

```python
# The first element is the column header which we'll need to remove to convert
# this into a single data type to become an np array and make sure to create
# the array as integers
incomes = incomes[1::]
np_incomes = np.array(incomes).astype(np.int32)
np_incomes
```

Now that we have all the income data from the mall customer csv file, let's find out more about those incomes using the np functions on np arrays.  Examine the code below to see how we're able to easily get back values like min, mean, median, and maximum

```python
print("average income: ",np.mean(np_incomes))
print("median income: ",np.median(np_incomes))
print("minimum income: ",np.min(np_incomes))
print("maximum income: ",np.max(np_incomes))
```

```python
# Age statistic
ages = [row[2] for row in my_data]
ages = ages[1::]
np_ages = np.array(ages).astype(np.int32)
print("average income: ",np.mean(np_ages))
print("median income: ",np.median(np_ages))
print("minimum income: ",np.min(np_ages))
print("maximum income: ",np.max(np_ages))
```

```python
# Correlation Cofficient for income vs spending score
# The Pearson product-moment correlation coefficient (or Pearson correlation coefficient, for short) is a measure of the strength of a linear association between two variables

incomes = [row[3] for row in my_data]
incomes = incomes[1::]
np_incomes = np.array(incomes).astype(np.int32)

spending_scores = [row[4] for row in my_data]
spending_scores = spending_scores[1::]
np_spending_scores = np.array(spending_scores).astype(np.int32)
```

```python
x_simple = np_incomes
y_simple = np_spending_scores
my_rho = np.corrcoef(x_simple, y_simple)
my_rho
```
