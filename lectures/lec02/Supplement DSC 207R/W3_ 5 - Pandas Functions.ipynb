# **Sort Values** -  sort_values() function sorts a data frame in Ascending or Descending order of passed Column.

```python
iris.sort_values(by = ["SepalLengthCm"])
```

```python
iris.sort_values(by = ["SepalLengthCm","SepalWidthCm"])
```

```python
iris.sort_values(by = ["PetalLengthCm"], ascending=False)
```
