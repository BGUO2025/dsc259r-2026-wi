```python
import numpy as np
```

```python
a = np.array([[1,2,3],[4,5,6]])
b = np.array([3,6,9])
```

```python
a
```

```python
a.shape
```

```python
b
```

```python
b.shape
```

```python
# can we add a and b?  This will be done via broadcasting.  b will be
# replicated to match the desired dimension
a+b
```

```python
# behind the scenes, b gets broadcasted into an array like this below
b_broadcasted = np.array([[3,6,9], [3,6,9]])
b_broadcasted
```

```python
a+b_broadcasted
```

Let's do an example of broadcasting for both arrays

```python
x = np.array([ [10,20,30] ])
x
```

```python
x.shape
```

```python
y = np.array([ [2], [4], [6] ])
y
```

```python
y.shape
```

```python
# if we add these, we'll get a 3x3 matrix.  Specifically, for the number of
# rows, we'll get the greater between 1 and 3 (so 3) and for the number of
# columns, we'll get the greater between 3 and 1 (so 3).  So both arrays
# will be broadcasted
x+y
```

```python
# you can think of x as being turned into a 3x3 matrix like this
x_b = np.array([ [10,20,30], [10,20,30], [10,20,30] ])
x_b
```

```python
# and y as a 3x3 matrix like this
y_b = np.array([ [2,2,2], [4,4,4], [6,6,6] ] )
y_b
```

```python
# so we get the same result as above
x_b+y_b
```

For more details on broadcasting rules, see:
