```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# Initialize the VADER sentiment analyzer
analyzer = SentimentIntensityAnalyzer()

# Sample texts for sentiment analysis
texts = [
    "I love this product! It's amazing.",
    "This is the worst experience I've ever had.",
    "I feel okay about the service.",
    "What a beautiful day!",
    "I'm not sure how I feel about this.",
    "This is absolutely fantastic!",
    "I hate this!",
    "It's not bad, but could be better.",
    "Great experience but poor floor mat.",
    "Poor floor mat but great experience."
]

# Analyze each text
for text in texts:
    scores = analyzer.polarity_scores(text)
    compound_score = scores['compound']
    sentiment = "Neutral"

    # Determine the sentiment category based on compound score
    if compound_score >= 0.05:
        sentiment = "Positive"
    elif compound_score <= -0.05:
        sentiment = "Negative"

    print(f"Text: {text}\nSentiment: {sentiment}, Scores: {scores}\n")
```
