<h3>Sentence & Word Tokenization In NLTK</h3>

```python
import nltk
from nltk.tokenize import sent_tokenize
nltk.download('punkt_tab')
```

```python
sent_tokenize("Dr. Strange loves the deep-dish pizza from Chicago. Hulk loves the cheesesteaks from Philadelphia.")
```

```python
from nltk.tokenize import word_tokenize
```

```python
word_tokenize("Dr. Strange loves the deep-dish pizza from Chicago. Hulk loves the cheesesteaks from Philadelphia.")
```

<h3>Sentence & Word Tokenization In Spacy</h3>

```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")

doc = nlp("Dr. Strange loves the deep-dish pizza from Chicago. Hulk loves the cheesesteaks from Philadelphia.")
```

```python
for sentence in doc.sents:
    print(sentence)
```

```python
for sentence in doc.sents:
    for word in sentence:
        print(word)
```
