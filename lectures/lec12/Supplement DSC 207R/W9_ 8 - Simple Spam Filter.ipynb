# Simple Spam Filter

```python
import pandas as pd
import numpy as np
```

```python

# Sample dataset
data = {
    'text': [
        'Win money now!', 'Hello, how are you?', 'Get cheap loans',
        'Call me when you can', 'Congratulations, you have won a prize!',
        'Meeting tomorrow at 10 am', 'Limited time offer, buy now!',
        'Hope you are doing well', 'Free entry in a cash prize draw',
        'Let\'s have lunch sometime'
    ],
    'label': ['spam', 'ham', 'spam', 'ham', 'spam', 'ham', 'spam', 'ham', 'spam', 'ham']
}
df = pd.DataFrame(data)
df
```

## Convert the text labels to numeric labels

```python
df['spam'] = df['label'].apply(lambda x: 1 if x =='spam' else 0)
df
```

## Train test split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['text'], df['spam'],  test_size=0.3)
```

```python
print(X_train)
print(X_test)
```

## Create bag of words representation using CountVectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()

X_train_vectorized = cv.fit_transform(X_train.values)
print(X_train_vectorized)
print(cv.get_feature_names_out())
```

## ASIDE: Understanding what happens in fit transform

```python
from sklearn.feature_extraction.text import CountVectorizer

# Sample text data
X = ["Win money now! Money money now!", "Hello, how are you now? hello?", "Get cheap loans now"]

# Initialize CountVectorizer
vectorizer = CountVectorizer()

# Convert text data into numerical features
X_vectorized = vectorizer.fit_transform(X)

# Print the feature names (words)
print(vectorizer.get_feature_names_out())

print(X_vectorized.toarray())


```

```python
pd.DataFrame(X_vectorized.toarray(), columns=vectorizer.get_feature_names_out(), index=X)
```

```python
# Print the resulting sparse matrix
print(X_vectorized)
```

## Continuing with our spam filter...

## Train the naive bayes model

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(X_train_vectorized, y_train)
```

```python
X_test_vectorized = cv.transform(X_test)
X_test_vectorized
```

## Evaluation Report


```python
from sklearn.metrics import classification_report

y_pred = model.predict(X_test_vectorized)

print(classification_report(y_test, y_pred))
```
