<h2 align="center">Tokenization in Spacy</h2>

```python
import spacy
```

Create blank language object and tokenize words in a sentence

```python
nlp = spacy.blank("en")

doc = nlp("Dr. Strange loves pav bhaji of mumbai as it costs only $2 per plate.")

for token in doc:
    print(token)
```

Creating blank language object gives a tokenizer and an empty pipeline. We will look more into language pipelines in next tutorial

<h3>Collecting email ids from text</h3>

```python
text = '''
Dear Team,

I hope this message finds you well. I wanted to remind everyone about the upcoming project deadlines and provide some important updates.

Firstly, the design phase of the project should be completed by July 25th. If you have any questions or need assistance, please reach out to our lead designer, John Doe, at john.doe@example.com. He will be available for consultations and can provide the necessary resources to ensure we stay on track.

Secondly, our marketing team is working on the launch strategy. For any marketing-related inquiries, you can contact Jane Smith at jane.smith@marketingcorp.com. Jane has extensive experience in this area and will be happy to assist with any promotional materials or strategies you may need.

Lastly, I would like to introduce our new team member, Emily Brown, who will be joining us as a software engineer. Emily can be reached at emily.brown@techsolutions.com. She brings a wealth of experience and is eager to contribute to our project.

Please make sure to include the respective team members in your communications if you have any specific questions or concerns.

Best regards,

Alex Johnson
Project Manager
alex.johnson@company.com
'''
text
```

```python
text = text.replace('\n', '')
text
```

```python
doc = nlp(text)
emails = []
for token in doc:
    if token.like_email:
        emails.append(token.text)
emails
```

<h3>Sentence Tokenization or Segmentation</h3>

```python
nlp = spacy.blank("en")
doc = nlp("Dr. Strange loves pav bhaji of mumbai. Hulk loves chat of delhi")
for sentence in doc.sents:
    print(sentence)
```

```python
nlp.pipeline
```

```python
nlp.add_pipe('sentencizer')
nlp.pipeline
```

```python
doc = nlp("Dr. Strange loves pav bhaji of mumbai. Hulk loves chat of delhi")
for sentence in doc.sents:
    print(sentence)
```

<h3>Exercise</h3>

(1) Think stats is a free book to study statistics (https://greenteapress.com/thinkstats2/thinkstats2.pdf)

This book has references to many websites from where you can download free datasets. You are an NLP engineer working for some company and you want to collect all dataset websites from this book. To keep this exercise simple you are given a paragraph from this book and you want to grab all urls from this paragraph using spacy

```python
text='''
Look for data to help you address the question. Governments are good
sources because data from public research is often freely available. Good
places to start include http://www.data.gov/, and http://www.science.
gov/, and in the United Kingdom, http://data.gov.uk/.
Two of my favorite data sets are the General Social Survey at http://www3.norc.org/gss+website/,
and the European Social Survey at http://www.europeansocialsurvey.org/.
'''


doc = nlp(text)
urls = []
for token in doc:
    if token.like_url:
        urls.append(token.text)
urls
```

(2) Extract all money transaction from below sentence along with currency. Output should be,

two $

500 €

```python
transactions = "Tony gave two $ to Peter, Bruce gave 500 € to Steve US worth thirty *"


doc = nlp(transactions)
for token in doc:
    if token.is_currency and doc[token.i - 1].like_num:
        print(doc[token.i - 1].text, token.text)

for token in doc:
    if token.like_num and doc[token.i + 1].is_currency:
        print(token.text, doc[token.i + 1].text)
```
