<h3>Stemming in NLTK</h3>

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

```python
words = ["eating", "eats", "adjustable", "rafting", "meeting", "running", "swimming",
         "ability", "children", "happily"]

for word in words:
    print(word, "|", stemmer.stem(word))
```

<h3>Lemmatization in Spacy</h3>

```python
import spacy
```

```python
import spacy

# Load the small English model
nlp = spacy.load("en_core_web_sm")

text = "The children were running quickly towards the playground, enjoying their playtime despite the windy weather."

doc = nlp(text)

for token in doc:
    print(token, " | ", token.lemma_, "|", token.lemma)
```

```python
doc = nlp("eating eats eat ate adjustable rafting ability meeting better")
for token in doc:
    print(token, " | ", token.lemma_)
```

<h3>Customizing lemmatizer</h3>

```python
nlp.pipe_names
```

```python
ar = nlp.get_pipe('attribute_ruler')

ar.add([[{"TEXT":"Bro"}],[{"TEXT":"Brah"}]],{"LEMMA":"Brother"})

doc = nlp("Bro, you wanna go? Brah, don't say no! I am exhausted")
for token in doc:
    print(token.text, "|", token.lemma_)
```
