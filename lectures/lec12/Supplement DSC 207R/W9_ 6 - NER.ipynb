# Named Entity Recognition (NER)

```python
import spacy

# Load the small English model
nlp = spacy.load("en_core_web_sm")

# Example text for Named Entity Recognition
text = "Barack Obama was born in Honolulu, Hawaii. He served as the 44th President of the United States from 2009 to 2017."

# Process the text using spaCy
doc = nlp(text)

# Extract and display named entities
for ent in doc.ents:
    print(f"Entity: {ent.text}, Label: {ent.label_}, Entity Type: {spacy.explain(ent.label_)}")
```

# Practical use case for NER

```python
import spacy

# Load the small English model
nlp = spacy.load("en_core_web_sm")

# Sample news article
text = """
President Joe Biden met with the Prime Minister of India, Narendra Modi, in New Delhi yesterday. They discussed global economic issues and climate change. They are planning to meet again in six months at Google in Mountain View, California.
"""

# Process the text using spaCy
doc = nlp(text)

# Extract named entities of interest
# TODO: WRITE CODE HERE
people = []
organizations = []
locations = []

for token in doc.ents:
    if token.label_ == "PERSON":
        people.append(token.text)
    elif token.label_ == "ORG":
        organizations.append(token.text)
    elif token.label_ == "GPE":
        locations.append(token.text)

# Print extracted information
print("People Mentioned:", people)
print("Organizations Mentioned:", organizations)
print("Locations Mentioned:", locations)
```
