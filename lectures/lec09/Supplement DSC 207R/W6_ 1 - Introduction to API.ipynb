### What is an API?

By definition from Wikipedia, "An application programming interface (API) is a connection between computers or between computer programs. It is a type of software interface, offering a service to other pieces of software."

WhatisAPI.png

In layman’s terms, it is a software intermediary that allows two applications to talk to each other. Think of an API as a translator between two people who don’t speak the same language but can communicate using a go-between.

### What is JSON?

##### JSON is a standard file format used for storing and transmitting data. Almost all APIs use JSON for sending and receiving text and numbers. XML is often used as well for its flexibility. However, in JSON we can send images and other types by encoding the data as text and numbers.

##### As seen in previous weeks, we know that CSV files are commonly used for processing data with Pandas and other libraries. APIs commonly retrieve data in JSON format, so we need to convert JSON data to CSV.

### How to make an API call with Python?

```python
import requests
import json
import pandas as pd

url = 'https://api.chucknorris.io/jokes/random'
response = requests.get(url)
string_response = response.content.decode("utf-8")
res = json.loads(string_response)
res
```
